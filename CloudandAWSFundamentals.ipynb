```python
# 1. Explain the differences between IaaS, PaaS, and SaaS. Provide one AWS example for each.
"""
IaaS: Provides basic infrastructure like virtual machines, storage, and networking. You manage the OS and applications.
 example: Amazon EC2
PaaS: Provides a platform to build and run applications without managing servers.
 example: AWS Elastic Beanstalk
SaaS: Provides ready-to-use software over the internet.
 example: Amazon QuickSight

"""
```

```python
# 2. Compare Public Cloud, Private Cloud, and Hybrid Cloud deployment models.
"""
Public Cloud: Cloud services are provided by companies like AWS and shared among different customers.
Private Cloud: Cloud infrastructure is dedicated to one organization.
Hybrid Cloud: Combines public cloud with private cloud or on-premises infrastructure.
example: A company may keep sensitive data on its private system and use AWS for other applications. This is a Hybrid Cloud.
"""
```

```python
# 3. What are AWS Regions and Availability Zones (AZs)? Why are they important?
"""
AWS Region: A geographical location where AWS has data centers, such as Mumbai or Singapore.
Availability Zone (AZ): An isolated data-center location inside an AWS Region.
They are important because they help provide high availability, better performance, and disaster recovery.
"""
```

```python
# 4. Identify AWS services used for compute, storage, networking, and databases.
"""
AWS provides different services for different purposes. Amazon EC2 and AWS Lambda are used for compute, while Amazon S3 and Amazon 
EBS are used for storage. Amazon VPC and Amazon Route 53 are used for networking. Amazon RDS and Amazon DynamoDB are used for managing 
databases.
"""
```

```python
#5. Differentiate between Infrastructure as Code (IaC), Containers, and Serverless Computing.
"""
IaC : Used to create and manage infrastructure using code.
Example: AWS CloudFormation
Containers: Package an application and its dependencies together so it can run consistently.
Example: Amazon ECS
Serverless: Allows you to run code without managing servers.
Example: AWS Lambda
"""
```

```python
#6. Compare AWS Management Console, AWS CLI, and AWSSDKs. When would you use each?
"""
AWS Management Console: A web-based interface where you manage AWS using clicks. Good for beginners and visual management.
AWS CLI: Allows you to manage AWS using commands. Useful for automation and quick tasks.
AWS SDKs: Programming libraries used to connect applications with AWS services. Useful when developing software."""
```
