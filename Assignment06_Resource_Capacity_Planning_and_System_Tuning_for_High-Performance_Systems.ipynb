```python
#1. How can you identify CPU and I/O bottlenecks in a Linux server
#using uptime and load averages?

"""
Run the uptime command to check the system load average.
Load averages represent the average number of processes waiting for CPU or I/O resources over the last 1, 5, and 15 minutes.
Compare the load average with the number of CPU cores.
If the load average is close to the number of CPU cores, the system is under normal load.
If it is consistently higher than the number of CPU cores, it may indicate a CPU or I/O bottleneck.
A high load average combined with low CPU utilization often suggests an I/O bottleneck (such as slow disk operations).

Example:
uptime
"""
```

```python
#2. How do you identify memory issues such as high swap usage and low free memory in Linux?

"""Memory issues can be identified using commands such as:

free -h
vmstat
top

Check for:

High swap usage (Swap Used)
Very low available memory (Available)
Increasing swap activity (si and so columns in vmstat)

High swap usage usually indicates that the system does not have enough physical RAM and is using disk space as virtual memory, which can reduce performance.
"""
```

```python
#3. What is the purpose of sar and Prometheus in system monitoring?
"""sar (System Activity Reporter)
Collects and reports historical system performance data.
Monitors CPU, memory, disk I/O, network, and other system resources.
Helps analyze performance over time and troubleshoot bottlenecks.

Example:

sar -u
Prometheus
An open-source monitoring and alerting tool.
Collects real-time metrics from servers and applications.
Stores time-series data.
Integrates with Grafana for dashboards and visualization.
Generates alerts when system metrics exceed defined thresholds."""
```

```python
#4. How can CPU usage be optimized in Linux systems?
"""
CPU usage can be optimized by:

Identifying high CPU-consuming processes using top or htop.
Stopping unnecessary background services.
Using nice and renice to adjust process priorities.
Optimizing application code and reducing unnecessary computations.
Upgrading hardware or adding more CPU cores if required.
Scheduling heavy jobs during off-peak hours.

Useful commands:

top
ps aux --sort=-%cpu
renice"""
```

```python
#5. How do you tune Linux kernel parameters using sysctl?
"""
The sysctl command is used to view and modify Linux kernel parameters at runtime.

View all parameters:

sysctl -a

View a specific parameter:

sysctl vm.swappiness

Temporarily change a parameter:

sudo sysctl -w vm.swappiness=10

To make the change permanent, edit:

/etc/sysctl.conf

or create a configuration file under:

/etc/sysctl.d/

Then reload the configuration:

sudo sysctl -p"""
```

```python
#6. Why are different disk partitioning strategies used for read-heavy and write-heavy workloads?
"""
Different workloads have different storage requirements.

Read-heavy workloads
Prioritize fast read performance.
Use SSDs, larger read caches, and optimized file systems.
Suitable for web servers, media servers, and databases with frequent reads.
Write-heavy workloads
Prioritize fast and reliable write operations.
Use RAID configurations, journaling file systems, and separate partitions for logs or databases.
Reduce disk fragmentation and improve write performance.
Suitable for database servers, logging systems, and transaction processing applications.

Using appropriate partitioning strategies improves system performance, reliability, and resource utilization."""
```
