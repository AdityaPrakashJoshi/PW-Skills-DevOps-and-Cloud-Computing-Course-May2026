```python
# 1. Explain the components of AWS Identity and Access Management (IAM): Users, Groups, Roles, and Policies.

"""
*AWS IAM is a service used to securely control  who can access AWS resources and what they are allowed to do.

Users : An IAM User represents a specific person or application that needs long-term access to AWS. A user can have credentials such 
as a password or access keys.

Groups : A Group is a collection of IAM users. Instead of assigning the same permissions to every user individually, we can attach 
policies to a group and add users to it.

Roles : A Role is an identity with permissions that can be **assumed temporarily**. Roles are commonly used by AWS services, applications, 
and users who need temporary access.

Policies : A Policy is a JSON document that defines **what actions are allowed or denied** on AWS resources.

"""
```

```python
# 2. What are MFA and Access Keys in AWS, and why are they important?
"""
MFA (Multi-Factor Authentication) : MFA adds an additional layer of security to an AWS account. Instead of relying only on a password, 
the user must provide another authentication factor, such as a code from an authenticator device or security key.
Even if an attacker obtains a password, MFA makes it much harder for them to access the account.

Access Keys : Access Key ID and a Secret Access Key—used to authenticate programmatic requests made to Amazon Web Services. They are 
vital because they allow external applications, scripts, and command-line tools to securely connect, automate, and manage cloud 
resources without a human browser login.
"""
```

```python
# 3. Describe AWS KMS, ACM, and WAF and their use cases.


"""
AWS KMS (Key Management Service)
AWS KMS is used to **create and manage encryption keys** that protect data.

Use cases:
- Encrypting S3 objects
- Encrypting EBS volumes
- Encrypting databases
- Controlling who can use encryption keys


AWS ACM (Certificate Manager)
AWS Certificate Manager is used to provision, manage, and deploy SSL/TLS certificates for applications and websites.

Use cases:
- Enable HTTPS for websites
- Secure traffic between users and applications
- Use certificates with services such as Elastic Load Balancing and CloudFront


AWS WAF (Web Application Firewall)
AWS WAF protects web applications from common web attacks by inspecting HTTP/HTTPS requests and applying rules.

Use cases:
- Block malicious IP addresses
- Protect against SQL injection
- Block cross-site scripting (XSS) patterns
- Rate-limit suspicious traffic
"""
```

```python
# 4. What is Amazon EC2? Explain instance types and Amazon Machine Images (AMIs).

"""
 Amazon EC2
Amazon EC2 (Elastic Compute Cloud) is an AWS service that provides virtual servers in the cloud. You can launch a server, 
configure it, run applications on it, and pay according to the pricing model you choose.

An EC2 instance provides resources such as:
- CPU
- Memory (RAM)
- Storage
- Networking

 EC2 Instance Types
Different workloads need different combinations of CPU, memory, storage, and networking resources. AWS provides different instance families for these needs.

- General Purpose: Balanced CPU, memory, and networking. Good for web servers and typical applications.
- Compute Optimized: More CPU power. Good for CPU-intensive workloads.
- Memory Optimized: More RAM. Good for databases and in-memory applications.
- Storage Optimized: Designed for workloads requiring high and fast local storage access.
- Accelerated Computing: Uses GPUs or other accelerators for workloads such as machine learning and graphics processing.
"""
```

```python
# 5. How do Auto Scaling and Elastic Load Balancing improve application availability?


"""Auto Scaling: Amazon EC2 Auto Scaling automatically adjusts the number of EC2 instances running an application according to demand and configured policies.

Example:
- Normal traffic → 2 EC2 instances
- Traffic increases → Auto Scaling launches more instances
- Traffic decreases → Auto Scaling can terminate unnecessary instances

This helps maintain performance while avoiding unnecessary infrastructure costs.

Elastic Load Balancing : Elastic Load Balancing distributes incoming application traffic across multiple targets, such as EC2 instances.
"""
```

```python
# 6. What is Serverless Computing, and how does AWS Lambda support it?
"""
Serverless Computing:
Serverless computing is a cloud model where the cloud provider manages the servers, infrastructure, scaling, and maintenance. The developer focuses mainly on writing and deploying code.

AWS Lambda:
AWS Lambda is a serverless compute service that runs code in response to events. AWS automatically manages the required infrastructure and scaling.

Common triggers: S3 events, API Gateway requests, EventBridge events, scheduled events, and messages from AWS services.

Benefits:

No server management
Automatic scaling
Pay mainly for code execution
Suitable for event-driven and short-running applications"""
```
