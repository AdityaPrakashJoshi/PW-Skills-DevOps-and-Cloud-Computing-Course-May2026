```python
#6. Explain the purpose of os, sys, subprocess, and shutil modules with examples.

""" 
1. os module: The os module provides a way to interact with the operating system. It allows you to perform tasks such as file and directory manipulation, environment variable access, and process management.

"""

print("Example of os module:")
import os
print(os.getcwd())


""" 
2. sys module: The sys module provides access to system-specific parameters and functions. It allows you to interact with the Python interpreter, handle command-line arguments, and manage the execution environment.
"""

print("\nExample of sys module:")
import sys
print(sys.argv)


""" 
3. subprocess module: The subprocess module allows you to spawn new processes, connect to their input/output/error pipes, and obtain their return codes. It is used for executing external commands and scripts from within Python."""

print("\nExample of subprocess module:")
import subprocess

subprocess.run(["echo", "Hello World"])

""" 
4. shutil module: The shutil module provides a higher-level interface for file operations, such as copying, moving, and deleting files and directories. It also includes functions for archiving and extracting files.
"""

print("\nExample of shutil module:")
import shutil

shutil.copy("file1.txt", "file2.txt")
```
