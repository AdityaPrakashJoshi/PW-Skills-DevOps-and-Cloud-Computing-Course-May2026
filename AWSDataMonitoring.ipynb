```python
#1. What is Amazon RDS, and what are the benefits of Multi-AZdeployments and Read Replicas?
"""
Amazon RDS (Relational Database Service) is a managed AWS service that makes it easier to set up, 
operate, and scale relational databases such as MySQL, PostgreSQL, and SQL Server.
- Multi-AZ deployment: Maintains a standby database in another Availability Zone to improve 
availability and support automatic failover if the primary database has a failure.
- Read Replicas: Create copies of a database to handle read requests, reducing the workload on 
the primary database and improving read performance.
"""
```

```python
#2. What is Amazon DynamoDB, and how is data organized in DynamoDB?
"""
Amazon DynamoDB is a fully managed NoSQL database service designed for fast, predictable performance 
at scale.
Data is organized into:
- Tables: Containers that store data.
- Items: Individual records in a table, similar to rows in a relational database.
- Attributes: Individual data fields within an item, similar to columns.
- Primary key: Uniquely identifies an item using a partition key, or a partition key combined with a 
sort key.
DynamoDB is schema-flexible, meaning items in the same table can have different attributes, provided 
they follow the table's key requirements.
Example: An online store can use a DynamoDB table to store customer details, with each customer 
identified by a unique customer ID.
"""
```

```python
#3. What are Metrics, Alarms, Dashboards, and Log Groups in Amazon CloudWatch?
"""

- Metrics: Numerical measurements collected over time, such as EC2 CPU utilization or request counts.
- Alarms: Monitor metrics and perform configured actions when specified thresholds or conditions 
are met.
- Dashboards: Visual displays that bring together metrics and alarms from different resources in one 
place.
- Log Groups: Containers that organize log streams, such as application logs or EC2-related logs, 
for monitoring and troubleshooting.
"""
```

```python
#4. What is AWS CloudTrail, and how does it help with auditing AWS accounts?

"""
AWS CloudTrail records account activity and API events, helping administrators understand who 
performed an action, what action was performed, and when it occurred.
It helps with security investigations, compliance auditing, and tracking changes to AWS resources.
"""
```
